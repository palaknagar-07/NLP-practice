```python
corpus = "Natural Language Processing (NLP) is a field of Artificial Intelligence. Tokenization is the process of breaking text into smaller units called tokensThese tokens can be words, sentences, or even characters. Students learn tokenization before moving to advanced NLP concepts like embeddings and transformers."
```

```python
# sent_tokenize breaks text into sentences.
# It finds sentence endings like periods and returns each sentence.

from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
for docs in documents:
    print(docs)
```

```python
from nltk.tokenize import word_tokenize
# word_tokenize breaks text into words and punctuation tokens.
# It gives each word and punctuation mark as a separate item.
words = word_tokenize(corpus)

for word in words:
    print(word)
```

```python
from nltk.tokenize import wordpunct_tokenize
# wordpunct_tokenize breaks text at punctuation marks and keeps punctuation as its own token.
# It is a simple tokenizer for words and punctuation.
words = wordpunct_tokenize(corpus)
for word in words:
    print(word)
```

```python
from nltk.tokenize import TreebankWordTokenizer
# TreebankWordTokenizer breaks text into words using Treebank rules.
# It is a more detailed tokenizer that handles punctuation and contractions well.
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```
