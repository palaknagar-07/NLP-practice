```python
import pandas as pd
messages = pd.read_csv("dataset/SMSSpamcollection.txt", sep='\t', names=["label", "message"])
messages
```

```python
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
import re
corpus = []
```

```python
for msg in messages['message']:
    review = re.sub('[^a-zA-Z]', ' ', msg)
    review = review.lower()
    corpus.append(review)
print(corpus[:5])    
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=300, binary=False) ## For Binary BoW --> binary = True
x = cv.fit_transform(corpus).toarray()
```
