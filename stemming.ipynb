```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized", "Congratulations", "fairly", "willingly"]
```

```python

from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
for word in words:
    print(word+"------>"+stemming.stem(word))

```

```python
from nltk.stem import RegexpStemmer
Reg_stemmer = RegexpStemmer('ing$|s$|ables$', min=4)
Reg_stemmer.stem('eating')
Reg_stemmer.stem("Congratulations")
Reg_stemmer.stem("Surprisingly")

for word in words:
    print(word+"------>"+Reg_stemmer.stem(word))
```

```python
from nltk.stem import SnowballStemmer
snow_stemmer = SnowballStemmer("english")
for word in words:
    print(word+"------>"+snow_stemmer.stem(word))
```
