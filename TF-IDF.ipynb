```python
import pandas as pd
messages = pd.read_csv("dataset/SMSSpamcollection.txt", sep='\t', names=["label", "message"])
messages
```

```python
import re
import nltk
```

```python
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmetizer = WordNetLemmatizer()
```

```python
import re
corpus = []
```

```python
for msg in messages['message']:
    review = re.sub('[^a-zA-Z]', ' ', msg)
    review = review.lower()
    review = review.split()
    review = [lemmetizer.lemmatize(word) for word in review if word not in stopwords.words('english')]
    review = ' '.join(review)
    corpus.append(review)
print(corpus[:5])
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf = TfidfVectorizer(max_features=100)
x = tf.fit_transform(corpus).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30, linewidth=100000, 
                    formatter=dict(float= lambda x: '%.3g'%x))
x[:5]
```

```python
tfidf = TfidfVectorizer(max_features=50, ngram_range=(2,2))
X = tfidf.fit_transform(corpus).toarray()
X[:5]
```
