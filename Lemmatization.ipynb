```python
from nltk.stem import WordNetLemmatizer
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized", "Congratulations", "fairly", "willingly", "goes"]
```

```python
import nltk
nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()
for word in words:
    print(word+"-------->"+lemmatizer.lemmatize(word, pos='v'))
```
